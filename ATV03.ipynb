```python
#Dicionário inicial
notas = {'Teo': 8.5, 'Cristiano': 8.9, 'Matson': 7.7}
```

```python
#Exibindo nota do aluno Matson
print("Nota do Matson:", notas['Matson'])
```

```python
#Adicionando uma nova aluna e sua nota ao dicionario
notas['Bruna'] = 8.7
print(notas)
```

```python
#Atualizando nota do aluno Teo
notas['Teo'] = 9.2
print(notas)
```

```python
#Verificando se o aluno Carlos está no dicionario
print("Carlos está no dicionário?", 'Carlos' in notas)
```

```python
#Deletando o aluno Matson do dicionario
del notas['Matson']
print(notas)
```

```python
#Adicionando novos alunos ao dicionario
notas['Lívia'] = 9.5
notas['João'] = 7.8
print(notas)
```

```python
#Mostrando todas as chaves
print("Alunos:", list(notas.keys()))
```

```python
#Mostrando todos os valores
print("Notas:", list(notas.values()))
```

```python
#Usando .get para buscar se Islla está no dicionario
print("Nota da Islla:", notas.get('Islla', 'Aluno não encontrado'))
```

```python
#usando .get para buscar um aluno
print("Nota da Bruna:", notas.get('Bruna'))
```
